```python
import numpy as np
import pandas as pd
np.random.seed(67)
```

```python
datos = np.random.normal(loc = 4.0, scale = 0.5, size = 50)
df_1 = pd.DataFrame({
    "datos" : datos
})
df_1
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm

# Extract the data
data_to_plot = df_1['datos']

# Create the figure and axes for the plot
plt.figure(figsize=(10, 6))

# Plot the histogram (frequency distribution)
sns.histplot(data_to_plot, bins=10, kde=False, stat='density', color='skyblue', label='Histograma')

# Plot the Kernel Density Estimate (KDE) for a smooth representation of the distribution
sns.kdeplot(data_to_plot, color='red', linestyle='-', linewidth=2, label='Estimación de Densidad de Kernel (KDE)')

# Calculate parameters for the normal distribution (mean and standard deviation)
mu, std = data_to_plot.mean(), data_to_plot.std()

# Generate x-values for the normal distribution curve
x = np.linspace(data_to_plot.min(), data_to_plot.max(), 100)

# Plot the Normal PDF
plt.plot(x, norm.pdf(x, mu, std), color='green', linestyle='--', linewidth=2, label=f'PDF Normal (μ={mu:.2f}, σ={std:.2f})')

# Add title and labels for clarity
plt.title('Distribución de ' + 'datos' + ' con PDF Normal y KDE', fontsize=16)
plt.xlabel('Valor', fontsize=12)
plt.ylabel('Densidad', fontsize=12)
plt.legend(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.6)

# Show the plot
plt.show()
```
