**Suma de Vectores**

```python
def sumar_vectores(A, B):
    columnas = len(A)

    if columnas != len(B):
        return "Error: Los vectores deben tener la misma dimensión para sumarse."

    else:
        try:
            resultado = []
            fila = []
            for j in range(columnas):
                fila.append(float(A[j]) + float(B[j]))
            resultado.append(fila)
            return resultado

        except ValueError:
            return "Debe ingresar solo números."
```

**Producto Punto**

```python
def producto_punto(A, B):
    columnas = len(A)

    if columnas != len(B):
        return "Error: Los vectores deben tener la misma dimensión."

    else:
        try:
            suma = 0
            for j in range(columnas):
                suma += float(A[j]) * float(B[j])
            return suma

        except ValueError:
            return "Debe ingresar solo números."
```

**Producto Cruz**

```python
def producto_cruz(A, B):
    if len(A) != 3 or len(B) != 3:
        return "Error: El producto cruz solo se define para vectores de dimensión 3."

    try:
        ax, ay, az = float(A[0]), float(A[1]), float(A[2])
        bx, by, bz = float(B[0]), float(B[1]), float(B[2])

        resultado = [ay*bz - az*by,
                     az*bx - ax*bz,
                     ax*by - ay*bx]
        return resultado

    except ValueError:
        return "Debe ingresar solo números."
```

**Producto de Hadamard entre vectores**

```python
def hadamard(A, B):
    columnas = len(A)

    if columnas != len(B):
        return "Error: Los vectores deben tener la misma dimensión."

    else:
        try:
            resultado = []
            fila = []
            for j in range(columnas):
                fila.append(float(A[j]) * float(B[j]))
            resultado.append(fila)
            return resultado

        except ValueError:
            return "Debe ingresar solo números."
```

**Menu De Opciones**

```python
v6 = input("Ingrese el vector 1 separado por espacios: ").split()
v7 = input("Ingrese el vector 2 separado por espacios: ").split()

opcion = 0

while opcion != 5:
    print("MENU DE OPCIONES")
    print("1. Suma de vectores")
    print("2. Producto punto")
    print("3. Producto cruz")
    print("4. Producto de Hadamard")
    print("5. Salir")

    try:
        opcion = int(input("Ingrese una opción: "))
    except ValueError:
        print("Debe ingresar un número.")
        continue

    if opcion == 1:
        r = sumar_vectores(v6, v7)
        print("La suma es:", r)

    elif opcion == 2:
        r = producto_punto(v6, v7)
        print("El producto punto es:", r)

    elif opcion == 3:
        r = producto_cruz(v6, v7)
        print("El producto cruz es:", r)

    elif opcion == 4:
        r = hadamard(v6, v7)
        print("El producto de Hadamard es:", r)
print("Fin del programa")
```
